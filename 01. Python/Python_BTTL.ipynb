```python
print("BTTL14")
import random

# a. Tạo mảng 2 chiều ngẫu nhiên
def tao_mang(m, n):
    mang = []
    for i in range(m):
        dong = []
        for j in range(n):
            dong.append(random.randint(1, 100))
        mang.append(dong)
    return mang

# In ma trận
def xuat_mang(mang):
    for dong in mang:
        for phan_tu in dong:
            print(phan_tu, end="\t")
        print()

# b. Xuất dòng k
def xuat_dong_k(mang, k):
    if k < 0 or k >= len(mang):
        print("Dòng không hợp lệ")
    else:
        print("Dòng", k, ":", mang[k])

# c. Xuất cột k
def xuat_cot_k(mang, k):
    if k < 0 or k >= len(mang[0]):
        print("Cột không hợp lệ")
    else:
        print("Cột", k, ":")
        for dong in mang:
            print(dong[k])

# d. Tìm dòng có tổng > 45 lớn nhất
def dong_tong_lon_nhat(mang):
    max_tong = -1
    vi_tri = -1
    for i in range(len(mang)):
        tong = sum(mang[i])
        if tong > 45 and tong > max_tong:
            max_tong = tong
            vi_tri = i
    if vi_tri == -1:
        print("Không có dòng thỏa điều kiện")
    else:
        print("Dòng", vi_tri, "có tổng lớn nhất:", max_tong)

# e. Tìm cột có tích nhỏ nhất
def cot_tich_nho_nhat(mang):
    n = len(mang[0])
    min_tich = None
    vi_tri = -1

    for j in range(n):
        tich = 1
        for i in range(len(mang)):
            tich *= mang[i][j]
        if min_tich is None or tich < min_tich:
            min_tich = tich
            vi_tri = j

    print("Cột", vi_tri, "có tích nhỏ nhất:", min_tich)

# f. Xuất phần tử dòng chẵn, cột lẻ
def dong_chan_cot_le(mang):
    print("Phần tử dòng chẵn, cột lẻ:")
    for i in range(len(mang)):
        for j in range(len(mang[0])):
            if i % 2 == 0 and j % 2 == 1:
                print(mang[i][j], end=" ")
    print()

# g. TBC phần tử chẵn thuộc dòng lẻ
def tbc_chan_dong_le(mang):
    tong = 0
    dem = 0
    for i in range(len(mang)):
        if i % 2 == 1:
            for j in range(len(mang[0])):
                if mang[i][j] % 2 == 0:
                    tong += mang[i][j]
                    dem += 1
    if dem == 0:
        print("Không có phần tử thỏa")
    else:
        print("TBC =", tong / dem)

# h. TBC các phần tử biên
def tbc_bien(mang):
    tong = 0
    dem = 0
    m = len(mang)
    n = len(mang[0])

    for i in range(m):
        for j in range(n):
            if i == 0 or i == m-1 or j == 0 or j == n-1:
                tong += mang[i][j]
                dem += 1
    print("TBC biên =", tong / dem)

# i. TBC tích các phần tử không thuộc biên
def tbc_tich_khong_bien(mang):
    tich = 1
    dem = 0
    m = len(mang)
    n = len(mang[0])

    for i in range(1, m-1):
        for j in range(1, n-1):
            tich *= mang[i][j]
            dem += 1

    if dem == 0:
        print("Không có phần tử")
    else:
        print("TBC tích =", tich ** (1/dem))

def menu():
    mang = []
    while True:
        print("\n===== MENU =====")
        print("1. Tạo mảng")
        print("2. Xuất dòng k")
        print("3. Xuất cột k")
        print("4. Dòng có tổng >45 lớn nhất")
        print("5. Cột có tích nhỏ nhất")
        print("6. Dòng chẵn, cột lẻ")
        print("7. TBC chẵn dòng lẻ")
        print("8. TBC phần tử biên")
        print("9. TBC tích không biên")
        print("0. Thoát")

        chon = int(input("Nhập lựa chọn: "))

        if chon == 1:
            m = int(input("Nhập số dòng: "))
            n = int(input("Nhập số cột: "))
            mang = tao_mang(m, n)
            xuat_mang(mang)

        elif chon == 2:
            k = int(input("Nhập dòng k: "))
            xuat_dong_k(mang, k)

        elif chon == 3:
            k = int(input("Nhập cột k: "))
            xuat_cot_k(mang, k)

        elif chon == 4:
            dong_tong_lon_nhat(mang)

        elif chon == 5:
            cot_tich_nho_nhat(mang)

        elif chon == 6:
            dong_chan_cot_le(mang)

        elif chon == 7:
            tbc_chan_dong_le(mang)

        elif chon == 8:
            tbc_bien(mang)

        elif chon == 9:
            tbc_tich_khong_bien(mang)

        elif chon == 0:
            break

        else:
            print("Chọn sai!")
menu()
```

```python
print("BTTL15")
# a. Class SinhVien
class SinhVien:
    def __init__(self, ma_sv, ten, nam_sinh, diem_tb):
        self.ma_sv = ma_sv
        self.ten = ten
        self.nam_sinh = nam_sinh
        self.diem_tb = diem_tb


# Nhập 1 sinh viên
def nhap_sinh_vien():
    ma_sv = input("Nhập mã SV (10 ký tự): ")
    ten = input("Nhập tên: ")
    nam_sinh = int(input("Nhập năm sinh: "))
    diem_tb = float(input("Nhập điểm TB: "))
    return SinhVien(ma_sv, ten, nam_sinh, diem_tb)


# b. Nhập danh sách sinh viên
def nhap_danh_sach():
    ds = []
    n = int(input("Nhập số lượng SV: "))
    for i in range(n):
        print(f"\nNhập SV thứ {i+1}:")
        sv = nhap_sinh_vien()
        ds.append(sv)
    return ds


# Xuất danh sách
def xuat_danh_sach(ds):
    for sv in ds:
        print(sv.ma_sv, "-", sv.ten, "-", sv.nam_sinh, "-", sv.diem_tb)


# c. Đếm SV đủ điều kiện lên lớp (>=5)
def dem_len_lop(ds):
    dem = 0
    for sv in ds:
        if sv.diem_tb >= 5:
            dem += 1
    print("Số SV lên lớp:", dem)


# d. Xuất SV đủ 20 tuổi
def sv_20_tuoi(ds):
    nam_hien_tai = 2026
    print("SV đủ 20 tuổi:")
    for sv in ds:
        if nam_hien_tai - sv.nam_sinh >= 20:
            print(sv.ma_sv, "-", sv.ten)


# e. Đếm SV hệ đại học (ký tự 2,3 là DH)
def dem_he_dh(ds):
    dem = 0
    for sv in ds:
        if sv.ma_sv[2:4] == "DH":
            dem += 1
    print("Số SV hệ DH:", dem)


# f. Đếm SV tên "Lan"
def dem_ten_lan(ds):
    dem = 0
    for sv in ds:
        if sv.ten.strip().split()[-1].lower() == "lan":
            dem += 1
    print("Số SV tên Lan:", dem)


# g. Đếm SV họ "Phan"
def dem_ho_phan(ds):
    dem = 0
    for sv in ds:
        if sv.ten.strip().split()[0].lower() == "phan":
            dem += 1
    print("Số SV họ Phan:", dem)

def menu():
    ds = []
    while True:
        print("\n===== MENU SINH VIÊN =====")
        print("1. Nhập danh sách SV")
        print("2. Xuất danh sách")
        print("3. Đếm SV lên lớp")
        print("4. Xuất SV đủ 20 tuổi")
        print("5. Đếm SV hệ DH")
        print("6. Đếm SV tên Lan")
        print("7. Đếm SV họ Phan")
        print("0. Thoát")

        chon = int(input("Chọn: "))

        if chon == 1:
            ds = nhap_danh_sach()

        elif chon == 2:
            xuat_danh_sach(ds)

        elif chon == 3:
            dem_len_lop(ds)

        elif chon == 4:
            sv_20_tuoi(ds)

        elif chon == 5:
            dem_he_dh(ds)

        elif chon == 6:
            dem_ten_lan(ds)

        elif chon == 7:
            dem_ho_phan(ds)

        elif chon == 0:
            break

        else:
            print("Chọn sai!")
menu()
```
