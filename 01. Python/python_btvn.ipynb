```python
import random
print("BTVN2")
# Tạo ma trận vuông n x n
def tao_ma_tran(n):
    a = []
    for i in range(n):
        dong = []
        for j in range(n):
            dong.append(random.randint(-50, 50))
        a.append(dong)
    return a

# Xuất ma trận
def xuat_ma_tran(a):
    for dong in a:
        for pt in dong:
            print(pt, end="\t")
        print()

# a. Tổng tam giác trên của đường chéo phụ (kể cả chéo phụ)
def tong_tam_giac_tren_cheo_phu(a):
    n = len(a)
    tong = 0
    for i in range(n):
        for j in range(n):
            if i + j <= n - 1:
                tong += a[i][j]
    print("Tổng =", tong)

# b. Đổi số âm thành trị tuyệt đối
def doi_so_am(a):
    n = len(a)
    for i in range(n):
        for j in range(n):
            if a[i][j] < 0:
                a[i][j] = abs(a[i][j])
    print("Đã đổi số âm thành dương")

# c. Thay số chẵn bằng x
def thay_so_chan(a):
    x = int(input("Nhập x: "))
    n = len(a)
    for i in range(n):
        for j in range(n):
            if a[i][j] % 2 == 0:
                a[i][j] = x
    print("Đã thay")

# d. Kiểm tra toàn chẵn
def kiem_tra_toan_chan(a):
    for dong in a:
        for pt in dong:
            if pt % 2 != 0:
                print("Không toàn chẵn")
                return
    print("Toàn chẵn")

# e. Kiểm tra đối xứng
def kiem_tra_doi_xung(a):
    n = len(a)
    for i in range(n):
        for j in range(n):
            if a[i][j] != a[j][i]:
                print("Không đối xứng")
                return
    print("Ma trận đối xứng")

# f. Kiểm tra chéo chính tăng dần
def cheo_chinh_tang(a):
    n = len(a)
    for i in range(n - 1):
        if a[i][i] > a[i+1][i+1]:
            print("Không tăng dần")
            return
    print("Chéo chính tăng dần")

# g. Xuất tam giác dưới của chéo phụ (kể cả chéo phụ)
def xuat_tam_giac_duoi_cheo_phu(a):
    n = len(a)
    print("Tam giác dưới chéo phụ:")
    for i in range(n):
        for j in range(n):
            if i + j >= n - 1:
                print(a[i][j], end=" ")
    print()

# h. Kiểm tra chéo phụ giảm dần
def cheo_phu_giam(a):
    n = len(a)
    for i in range(n - 1):
        if a[i][n-1-i] < a[i+1][n-2-i]:
            print("Không giảm dần")
            return
    print("Chéo phụ giảm dần")

def menu():
    a = []
    while True:
        print("\n===== MENU MA TRẬN =====")
        print("1. Tạo ma trận")
        print("2. Tổng tam giác trên chéo phụ")
        print("3. Đổi số âm")
        print("4. Thay số chẵn bằng x")
        print("5. Kiểm tra toàn chẵn")
        print("6. Kiểm tra đối xứng")
        print("7. Chéo chính tăng dần")
        print("8. Xuất tam giác dưới chéo phụ")
        print("9. Chéo phụ giảm dần")
        print("0. Thoát")

        chon = int(input("Chọn: "))

        if chon == 1:
            n = int(input("Nhập n: "))
            a = tao_ma_tran(n)
            xuat_ma_tran(a)

        elif chon == 2:
            tong_tam_giac_tren_cheo_phu(a)

        elif chon == 3:
            doi_so_am(a)
            xuat_ma_tran(a)

        elif chon == 4:
            thay_so_chan(a)
            xuat_ma_tran(a)

        elif chon == 5:
            kiem_tra_toan_chan(a)

        elif chon == 6:
            kiem_tra_doi_xung(a)

        elif chon == 7:
            cheo_chinh_tang(a)

        elif chon == 8:
            xuat_tam_giac_duoi_cheo_phu(a)

        elif chon == 9:
            cheo_phu_giam(a)

        elif chon == 0:
            break

        else:
            print("Chọn sai!")

menu()
```
