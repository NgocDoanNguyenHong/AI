```python
import copy
from heapq import heappush, heappop
print("Bài tập mẫu")
# Kích thước puzzle
n = 3

# Hướng di chuyển: xuống, trái, lên, phải
hang = [1, 0, -1, 0]
cot = [0, -1, 0, 1]

# ========================
# Hàng đợi ưu tiên
# ========================
class HangDoiUuTien:
    def __init__(self):
        self.heap = []

    def them(self, node):
        heappush(self.heap, node)

    def lay(self):
        return heappop(self.heap)

    def rong(self):
        return len(self.heap) == 0


# ========================
# Node
# ========================
class Nut:
    def __init__(self, cha, ma_tran, vi_tri_trong, chi_phi_h, muc):
        self.cha = cha
        self.ma_tran = ma_tran
        self.vi_tri_trong = vi_tri_trong
        self.chi_phi_h = chi_phi_h
        self.muc = muc  # g(n)

    def __lt__(self, nut_khac):
        return (self.chi_phi_h + self.muc) < (nut_khac.chi_phi_h + nut_khac.muc)
# ========================
# Tính heuristic (Manhattan)
# ========================
def tinh_heuristic(a, dich):
    khoang_cach = 0
    for i in range(n):
        for j in range(n):
            if a[i][j] != 0:
                for x in range(n):
                    for y in range(n):
                        if dich[x][y] == a[i][j]:
                            khoang_cach += abs(x - i) + abs(y - j)
    return khoang_cach
# ========================
# Kiểm tra hợp lệ
# ========================
def hop_le(x, y):
    return 0 <= x < n and 0 <= y < n
# ========================
# Tạo node mới
# ========================
def tao_nut(cha, ma_tran, x, y, new_x, new_y, muc, dich):
    mat_moi = copy.deepcopy(ma_tran)

    # đổi chỗ
    mat_moi[x][y], mat_moi[new_x][new_y] = mat_moi[new_x][new_y], mat_moi[x][y]

    chi_phi = tinh_heuristic(mat_moi, dich)

    return Nut(cha, mat_moi, (new_x, new_y), chi_phi, muc)

def in_loi_giai(nut):
    if nut is None:
        return
    in_loi_giai(nut.cha)
    for dong in nut.ma_tran:
        print(dong)
    print()
# ========================
# A* chính
# ========================
def giai_8_puzzle(bat_dau, dich, vi_tri_trong):
    hang_doi = HangDoiUuTien()

    h = tinh_heuristic(bat_dau, dich)
    goc = Nut(None, bat_dau, vi_tri_trong, h, 0)

    hang_doi.them(goc)

    while not hang_doi.rong():
        nut_hien_tai = hang_doi.lay()

        if nut_hien_tai.chi_phi_h == 0:
            print("Đã tìm thấy lời giải:\n")
            in_loi_giai(nut_hien_tai)
            return

        x, y = nut_hien_tai.vi_tri_trong

        for i in range(4):
            new_x = x + hang[i]
            new_y = y + cot[i]

            if hop_le(new_x, new_y):
                nut_moi = tao_nut(
                    nut_hien_tai,
                    nut_hien_tai.ma_tran,
                    x, y,
                    new_x, new_y,
                    nut_hien_tai.muc + 1,
                    dich
                )
                hang_doi.them(nut_moi)
# ========================
# Chạy thử
# ========================
bat_dau = [
    [1, 2, 3],
    [5, 6, 0],
    [7, 8, 4]
]
dich = [
    [1, 2, 3],
    [5, 8, 6],
    [0, 7, 4]
]

giai_8_puzzle(bat_dau, dich, (1, 2))
```
