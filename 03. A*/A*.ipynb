```python
import heapq

# Thuật toán A*
def a_sao(do_thi, bat_dau, dich, heuristic):
    hang_doi = []
    heapq.heappush(hang_doi, (0, bat_dau))

    # Lưu chi phí từ điểm bắt đầu
    chi_phi_g = {bat_dau: 0}

    # Lưu đường đi
    cha = {bat_dau: None}

    while hang_doi:
        f, dinh_hien_tai = heapq.heappop(hang_doi)

        if dinh_hien_tai == dich:
            return tai_tao_duong_di(cha, dich)

        for hang_xom, trong_so in do_thi[dinh_hien_tai]:
            g_moi = chi_phi_g[dinh_hien_tai] + trong_so

            if hang_xom not in chi_phi_g or g_moi < chi_phi_g[hang_xom]:
                chi_phi_g[hang_xom] = g_moi
                f_moi = g_moi + heuristic(hang_xom, dich)
                heapq.heappush(hang_doi, (f_moi, hang_xom))
                cha[hang_xom] = dinh_hien_tai
    return None

# Tái tạo đường đi
def tai_tao_duong_di(cha, dich):
    duong_di = []
    hien_tai = dich
    while hien_tai is not None:
        duong_di.append(hien_tai)
        hien_tai = cha[hien_tai]
    return duong_di[::-1]
# Ví dụ heuristic (khoảng cách giả định)
def heuristic(dinh, dich):
    # Ví dụ đơn giản (có thể thay bằng khoảng cách thực)
    bang_heuristic = {
        'A': 7,
        'B': 6,
        'C': 2,
        'D': 1,
        'E': 0
    }
    return bang_heuristic.get(dinh, 0)
# Đồ thị dạng danh sách kề
do_thi = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 5), ('E', 12)],
    'C': [('D', 2)],
    'D': [('E', 3)],
    'E': []
}
# Test
duong_di = a_sao(do_thi, 'A', 'E', heuristic)

print("Đường đi:", duong_di)
```

```python
import heapq

# Tính heuristic (ước lượng chi phí còn lại)
def tinh_heuristic(dinh_hien_tai, chua_tham, ma_tran_kc):
    if not chua_tham:
        return 0

    # Lấy cạnh nhỏ nhất từ đỉnh hiện tại đến các đỉnh chưa thăm
    nho_nhat = min(ma_tran_kc[dinh_hien_tai][i] for i in chua_tham)

    # Cộng thêm cạnh nhỏ nhất giữa các đỉnh chưa thăm
    tong = nho_nhat
    for i in chua_tham:
        min_canh = min(ma_tran_kc[i][j] for j in range(len(ma_tran_kc)) if i != j)
        tong += min_canh

    return tong

# A* giải TSP
def a_sao_tsp(ma_tran_kc):
    so_dinh = len(ma_tran_kc)
    bat_dau = 0

    # (f, g, dinh_hien_tai, da_tham, duong_di)
    hang_doi = []
    heapq.heappush(hang_doi, (0, 0, bat_dau, frozenset([bat_dau]), [bat_dau]))

    while hang_doi:
        f, g, dinh_hien_tai, da_tham, duong_di = heapq.heappop(hang_doi)

        # Nếu đã đi hết các đỉnh → quay về điểm đầu
        if len(da_tham) == so_dinh:
            chi_phi = g + ma_tran_kc[dinh_hien_tai][bat_dau]
            return duong_di + [bat_dau], chi_phi

        # Các đỉnh chưa thăm
        chua_tham = set(range(so_dinh)) - da_tham

        for dinh_ke in chua_tham:
            g_moi = g + ma_tran_kc[dinh_hien_tai][dinh_ke]
            da_tham_moi = da_tham | {dinh_ke}
            duong_di_moi = duong_di + [dinh_ke]

            h = tinh_heuristic(dinh_ke, chua_tham - {dinh_ke}, ma_tran_kc)
            f_moi = g_moi + h

            heapq.heappush(
                hang_doi,
                (f_moi, g_moi, dinh_ke, da_tham_moi, duong_di_moi)
            )

    return None, float('inf')


# Ví dụ ma trận khoảng cách
ma_tran_kc = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

duong_di, chi_phi = a_sao_tsp(ma_tran_kc)

print("Đường đi tối ưu:", duong_di)
print("Chi phí:", chi_phi)
```
