```python
print("Bài 1")
import copy
import math

X = "X"
O = "O"
RONG = None

def trang_thai_ban_dau():
    return [
        [RONG, RONG, RONG],
        [RONG, RONG, RONG],
        [RONG, RONG, RONG]
    ]

def nguoi_choi(bang):
    dem_x = sum(hang.count(X) for hang in bang)
    dem_o = sum(hang.count(O) for hang in bang)
    if dem_x == dem_o:
        return X
    return O

def nuoc_di_hop_le(bang):
    tap_hop = set()
    for i in range(3):
        for j in range(3):
            if bang[i][j] is RONG:
                tap_hop.add((i, j))
    return tap_hop

def ket_qua(bang, hanh_dong):
    i, j = hanh_dong
    bang_moi = copy.deepcopy(bang)
    bang_moi[i][j] = nguoi_choi(bang)
    return bang_moi

def nguoi_thang(bang):
    for hang in bang:
        if hang[0] == hang[1] == hang[2] and hang[0] is not None:
            return hang[0]

    for cot in range(3):
        if bang[0][cot] == bang[1][cot] == bang[2][cot] and bang[0][cot] is not None:
            return bang[0][cot]

    if bang[0][0] == bang[1][1] == bang[2][2] and bang[0][0] is not None:
        return bang[0][0]

    if bang[0][2] == bang[1][1] == bang[2][0] and bang[0][2] is not None:
        return bang[0][2]

    return None

def ket_thuc(bang):
    if nguoi_thang(bang) is not None:
        return True
    for hang in bang:
        if RONG in hang:
            return False
    return True

def diem_danh_gia(bang):
    thang = nguoi_thang(bang)
    if thang == X:
        return 1
    if thang == O:
        return -1
    return 0

def gia_tri_lon_nhat(trang_thai):
    if ket_thuc(trang_thai):
        return diem_danh_gia(trang_thai)

    gia_tri = -math.inf
    for hanh_dong in nuoc_di_hop_le(trang_thai):
        gia_tri = max(gia_tri, gia_tri_nho_nhat(ket_qua(trang_thai, hanh_dong)))
    return gia_tri

def gia_tri_nho_nhat(trang_thai):
    if ket_thuc(trang_thai):
        return diem_danh_gia(trang_thai)

    gia_tri = math.inf
    for hanh_dong in nuoc_di_hop_le(trang_thai):
        gia_tri = min(gia_tri, gia_tri_lon_nhat(ket_qua(trang_thai, hanh_dong)))
    return gia_tri

def minimax(bang):
    if ket_thuc(bang):
        return None

    luot = nguoi_choi(bang)

    if luot == X:
        gia_tri_tot_nhat = -math.inf
        nuoc_di_tot_nhat = None
        for hanh_dong in nuoc_di_hop_le(bang):
            gia_tri = gia_tri_nho_nhat(ket_qua(bang, hanh_dong))
            if gia_tri > gia_tri_tot_nhat:
                gia_tri_tot_nhat = gia_tri
                nuoc_di_tot_nhat = hanh_dong
        return nuoc_di_tot_nhat
    else:
        gia_tri_tot_nhat = math.inf
        nuoc_di_tot_nhat = None
        for hanh_dong in nuoc_di_hop_le(bang):
            gia_tri = gia_tri_lon_nhat(ket_qua(bang, hanh_dong))
            if gia_tri < gia_tri_tot_nhat:
                gia_tri_tot_nhat = gia_tri
                nuoc_di_tot_nhat = hanh_dong
        return nuoc_di_tot_nhat

def in_bang(bang):
    print("  0 1 2")
    for i, hang in enumerate(bang):
        dong = str(i) + " "
        for o in hang:
            if o is None:
                dong += ". "
            else:
                dong += o + " "
        print(dong)

def choi():
    bang = trang_thai_ban_dau()

    while not ket_thuc(bang):
        in_bang(bang)
        if nguoi_choi(bang) == X:
            while True:
                try:
                    user_input = input("Nhap hang cot (vd: 0 1): ").split()
                    if len(user_input) == 2:
                        hang, cot = map(int, user_input)
                        if (hang, cot) in nuoc_di_hop_le(bang):
                            bang = ket_qua(bang, (hang, cot))
                            break
                        else:
                            print("Nuoc di khong hop le. O nay da duoc chon hoac ngoai pham vi.")
                    else:
                        print("Dinh dang dau vao khong hop le. Vui long nhap 2 so cach nhau boi khoang trang.")
                except ValueError:
                    print("Dinh dang dau vao khong hop le. Vui long nhap 2 so nguyen cach nhau boi khoang trang.")
        else:
            nuoc = minimax(bang)
            bang = ket_qua(bang, nuoc)

    in_bang(bang)
    thang = nguoi_thang(bang)
    if thang:
        print("Thang:", thang)
    else:
        print("Hoa")
choi()
```
