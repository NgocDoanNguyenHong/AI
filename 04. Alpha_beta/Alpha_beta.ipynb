```python
print("Bài 2")
import copy
import math

X = "X"
O = "O"
RONG = None

def trang_thai_ban_dau():
    return [
        [RONG, RONG, RONG],
        [RONG, RONG, RONG],
        [RONG, RONG, RONG]
    ]

def nguoi_choi(bang):
    dem_x = sum(hang.count(X) for hang in bang)
    dem_o = sum(hang.count(O) for hang in bang)
    return X if dem_x == dem_o else O

def nuoc_di_hop_le(bang):
    tap = set()
    for i in range(3):
        for j in range(3):
            if bang[i][j] is RONG:
                tap.add((i, j))
    return tap

def ket_qua(bang, hanh_dong):
    i, j = hanh_dong
    bang_moi = copy.deepcopy(bang)
    bang_moi[i][j] = nguoi_choi(bang)
    return bang_moi

def nguoi_thang(bang):
    for hang in bang:
        if hang[0] == hang[1] == hang[2] and hang[0] is not None:
            return hang[0]

    for cot in range(3):
        if bang[0][cot] == bang[1][cot] == bang[2][cot] and bang[0][cot] is not None:
            return bang[0][cot]

    if bang[0][0] == bang[1][1] == bang[2][2] and bang[0][0] is not None:
        return bang[0][0]

    if bang[0][2] == bang[1][1] == bang[2][0] and bang[0][2] is not None:
        return bang[0][2]

    return None

def ket_thuc(bang):
    if nguoi_thang(bang) is not None:
        return True
    for hang in bang:
        if RONG in hang:
            return False
    return True

def diem(bang, do_sau):
    thang = nguoi_thang(bang)
    if thang == X:
        return 10 - do_sau
    if thang == O:
        return -10 + do_sau
    return 0

def alpha_beta(bang, do_sau, alpha, beta, la_max):
    if ket_thuc(bang):
        return diem(bang, do_sau)

    if la_max:
        gia_tri = -math.inf
        for hanh_dong in nuoc_di_hop_le(bang):
            gia_tri = max(
                gia_tri,
                alpha_beta(ket_qua(bang, hanh_dong), do_sau + 1, alpha, beta, False)
            )
            alpha = max(alpha, gia_tri)
            if beta <= alpha:
                break
        return gia_tri
    else:
        gia_tri = math.inf
        for hanh_dong in nuoc_di_hop_le(bang):
            gia_tri = min(
                gia_tri,
                alpha_beta(ket_qua(bang, hanh_dong), do_sau + 1, alpha, beta, True)
            )
            beta = min(beta, gia_tri)
            if beta <= alpha:
                break
        return gia_tri

def tim_nuoc_tot_nhat(bang):
    luot = nguoi_choi(bang)

    if luot == X:
        gia_tri_tot = -math.inf
        nuoc_tot = None
        for hanh_dong in nuoc_di_hop_le(bang):
            gia_tri = alpha_beta(
                ket_qua(bang, hanh_dong),
                0,
                -math.inf,
                math.inf,
                False
            )
            if gia_tri > gia_tri_tot:
                gia_tri_tot = gia_tri
                nuoc_tot = hanh_dong
    else:
        gia_tri_tot = math.inf
        nuoc_tot = None
        for hanh_dong in nuoc_di_hop_le(bang):
            gia_tri = alpha_beta(
                ket_qua(bang, hanh_dong),
                0,
                -math.inf,
                math.inf,
                True
            )
            if gia_tri < gia_tri_tot:
                gia_tri_tot = gia_tri
                nuoc_tot = hanh_dong

    return nuoc_tot

def in_bang(bang):
    print("  0 1 2")
    for i, hang in enumerate(bang):
        dong = str(i) + " "
        for o in hang:
            dong += (o if o else ".") + " "
        print(dong)

def choi():
    bang = trang_thai_ban_dau()

    while not ket_thuc(bang):
        in_bang(bang)

        if nguoi_choi(bang) == X:
            hang, cot = map(int, input("Nhap hang cot: ").split())
            if (hang, cot) in nuoc_di_hop_le(bang):
                bang = ket_qua(bang, (hang, cot))
        else:
            nuoc = tim_nuoc_tot_nhat(bang)
            bang = ket_qua(bang, nuoc)

    in_bang(bang)
    thang = nguoi_thang(bang)
    if thang:
        print("Thang:", thang)
    else:
        print("Hoa")

choi()
```
